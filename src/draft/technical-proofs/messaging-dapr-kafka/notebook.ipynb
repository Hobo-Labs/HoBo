# Messaging + Dapr + Kafka (ADR-0138)

Exercises Kafka and Dapr adapter paths with in-memory transports.

```typescript
import { strict as assert } from "node:assert";
import { runDaprKafkaPath, runNativeKafkaPath } from "./src/scenarios.js";
const native = await runNativeKafkaPath("demo");
const dapr = await runDaprKafkaPath("demo");
assert.equal(native.published, 2);
assert.equal(native.consumed, 2);
assert.equal(dapr.published, 1);
assert.equal(dapr.consumed, 1);
console.log(JSON.stringify({ transport: "in-memory", native: { published: native.published, consumed: native.consumed }, dapr: { published: dapr.published, consumed: dapr.consumed } }));
```
