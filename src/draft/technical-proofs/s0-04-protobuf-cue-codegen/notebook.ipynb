# Protobuf and CUE codegen check

Run the proof CLI against its protobuf and CUE fixtures.


```typescript
const { runCli } = await import("./packages/schema-tools/src/cli.ts");
const result = await runCli(["--check", "--json"], process.cwd());
if (result.exitCode !== 0) throw new Error(result.stderr || result.stdout);
const report = JSON.parse(result.stdout);
console.log(`status: ${report.status}; drift: ${report.drift.length}; cue violations: ${report.cueViolations.length}`);
```
